# for outliers removing

```python
q1=s['age'].quantile(0.25)
q3=s['age'].quantile(0.75)
iqr=q3-q1
upper_bound=q3+1.5*iqr
lower_bound=q1-1.5*iqr
s['age']=np.where(s['age']>upper_bound,upper_bound,s['age'])
s['age']=np.where(s['age']<lower_bound,lower_bound,s['age'])
```

```python
sns.boxplot(x=s['age'])
```

```python
upper kiya gya hai iska box plot to age me outliers the jo ab remove ho chuke hai 
```

```python
plt.figure(figsize=(12,6))
sns.boxplot(data=s,x="points",color="pink")
plt.show()
```

```python
q1=s['points'].quantile(0.25)
q3=s['points'].quantile(0.75)
iqr=q3-q1
upper_bound=q3+1.5*iqr
lower_bound=q1-1.5*iqr
s['points']=np.where(s['points']>upper_bound,upper_bound,s['points'])
s['points']=np.where(s['points']<lower_bound,lower_bound,s['points'])
```

```python
sns.boxplot(data=s,x="points",color="pink")
plt.show()
```
