```python
##8
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings 
warnings.filterwarnings('ignore')
```

```python
df = pd.read_csv('winequality-red.csv',sep=";")
df.head()
```

```python
df.shape
```

```python
len(df[df.duplicated()])
```

```python
df.isnull().sum()
```

```python
plt.figure(figsize=(25,8))
df["quality"].value_counts().plot.bar()
plt.show()
```

```python
df.corr(numeric_only=True)
```

```python
plt.figure(figsize=(15,10))
sns.heatmap(df.corr(numeric_only=True), annot=True, linewidths=0.5)

plt.title("Correlation Heatmap of DataFrame")
plt.show()
```

fixed acidity and dencity columns have highest correlation

```python
plt.figure(figsize=(20,8))
sns.boxplot(data=df,x='fixed acidity')
plt.show()
```

```python
q1=df['fixed acidity'].quantile(0.25)
q3=df['fixed acidity'].quantile(0.75)
iqr=q3-q1
upper_bound=q3+1.5*iqr
df['fixed acidity']=np.where(df['fixed acidity']>upper_bound,upper_bound,df['fixed acidity'])
```

```python
plt.figure(figsize=(20,8))
sns.boxplot(data=df,x='fixed acidity')
plt.show()
```

```python
plt.figure(figsize=(20,8)) 
sns.boxplot(data=df,x='total sulfur dioxide')
plt.show()
```

```python
q1=df['total sulfur dioxide'].quantile(0.25)
q3=df['total sulfur dioxide'].quantile(0.75)
iqr=q3-q1
upper_bound=q3+1.5*iqr
df['total sulfur dioxide']=np.where(df['total sulfur dioxide']>upper_bound,upper_bound,df['total sulfur dioxide'])
```

```python
plt.figure(figsize=(20,8)) 
sns.boxplot(data=df,x='total sulfur dioxide')
plt.show()
```

```python
plt.figure(figsize=(20,8)) 
sns.boxplot(data=df,x='density')
plt.show()
```

```python
q1=df['density'].quantile(0.25)
q3=df['density'].quantile(0.75)
iqr=q3-q1
upper_bound=q3+1.5*iqr
lower_bound=q1-1.5*iqr
df['density']=np.where(df['density']>upper_bound,upper_bound,df['density'])
df['density']=np.where(df['density']<lower_bound,upper_bound,df['density'])
```

```python
plt.figure(figsize=(20,8)) 
sns.boxplot(data=df,x='density')
plt.show()
```

```python
plt.figure(figsize=(20,8)) 
sns.boxplot(data=df,x='pH')
plt.show()
```

```python
q1=df['pH'].quantile(0.25)
q3=df['pH'].quantile(0.75)
iqr=q3-q1
upper_bound=q3+1.5*iqr
lower_bound=q1-1.5*iqr
df['pH']=np.where(df['pH']>upper_bound,upper_bound,df['pH'])
df['pH']=np.where(df['pH']<lower_bound,upper_bound,df['pH'])
```

```python
plt.figure(figsize=(20,8)) 
sns.boxplot(data=df,x='pH')
plt.show()
```
